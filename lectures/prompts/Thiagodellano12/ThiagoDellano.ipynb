# Manipulação de Arquivos e Serialização de Objetos em Java

## 📁 Manipulação Básica de Arquivos (Create, Read, Write, Update)

### 📝 Criação e Escrita com `FileWriter`

#### 1. O que é a classe `FileWriter`?

A classe `FileWriter` é usada para gravar caracteres em arquivos de texto. Ela pode criar um novo arquivo ou sobrescrever um existente. Também é possível configurar para acrescentar (`append`) texto sem apagar o conteúdo anterior.

#### 2. Exemplo - Criar e escrever no arquivo `dados.txt`

```java
import java.io.FileWriter;
import java.io.IOException;

public class CriarArquivo {
    public static void main(String[] args) {
        try (FileWriter writer = new FileWriter("dados.txt")) {
            writer.write("Primeira linha\n");
            writer.write("Segunda linha\n");
            writer.write("Terceira linha\n");
        } catch (IOException e) {
            e.printStackTrace();
        }
    }
}
```

#### 3. O que acontece se executar duas vezes?

Se não usar o modo de *append*, o conteúdo anterior será **sobrescrito**.

### 📖 Leitura com `BufferedReader`

#### 1. Uso de `FileReader` e `BufferedReader`

* `FileReader` lê arquivos texto caractere por caractere.
* `BufferedReader` adiciona uma camada de eficiência, lendo linhas inteiras.

#### 2. Exemplo de leitura

```java
import java.io.BufferedReader;
import java.io.FileReader;
import java.io.IOException;

public class LerArquivo {
    public static void main(String[] args) {
        try (BufferedReader reader = new BufferedReader(new FileReader("dados.txt"))) {
            String linha;
            while ((linha = reader.readLine()) != null) {
                System.out.println(linha);
            }
        } catch (IOException e) {
            e.printStackTrace();
        }
    }
}
```

#### 3. Como lidar com exceções?

Use `try-catch` com `IOException` para tratar erros de leitura.

### 🔁 Atualização de Arquivos (read → modify → write)

#### 1. Como atualizar arquivos?

É necessário:

* Ler o conteúdo,
* Modificá-lo em memória,
* Escrevê-lo novamente em um novo arquivo (ou sobrescrever o original).

#### 2. Exemplo: Substituir "teste" por "exemplo"

```java
import java.io.*;

public class AtualizarArquivo {
    public static void main(String[] args) {
        try (BufferedReader reader = new BufferedReader(new FileReader("dados.txt"));
             FileWriter writer = new FileWriter("dados_atualizado.txt")) {

            String linha;
            while ((linha = reader.readLine()) != null) {
                linha = linha.replaceAll("teste", "exemplo");
                writer.write(linha + "\n");
            }

        } catch (IOException e) {
            e.printStackTrace();
        }
    }
}
```

### 📁 Listagem de Diretórios com `DirectoryStream`

#### 1. Como listar arquivos?

Use `Files.newDirectoryStream()` com filtros.

#### 2. Exemplo: listar `.txt`

```java
import java.nio.file.*;
import java.io.IOException;

public class ListarTxt {
    public static void main(String[] args) throws IOException {
        Path dir = Paths.get(".");
        try (DirectoryStream<Path> stream = Files.newDirectoryStream(dir, "*.txt")) {
            for (Path file : stream) {
                System.out.println(file.getFileName());
            }
        }
    }
}
```

1. **Texto**: Conteúdo legível (ex: `.txt`, `.csv`).
2. **Binário**: Dados em forma compacta (ex: `.dat`, `.exe`, `.jpg`).
3. **Eficiência**: Binários são mais rápidos e compactos para objetos/dados estruturados.

### 💾 Leitura/Escrita com `DataInputStream` e `DataOutputStream`

#### 1. Funcionamento

* `DataOutputStream`: grava tipos primitivos.
* `DataInputStream`: lê os mesmos tipos.

#### 2. Gravar e ler inteiros

```java
// Gravar
try (DataOutputStream out = new DataOutputStream(new FileOutputStream("valores.dat"))) {
    for (int i = 0; i < 5; i++) {
        out.writeInt(i * 10);
    }
}

// Ler
try (DataInputStream in = new DataInputStream(new FileInputStream("valores.dat"))) {
    while (in.available() > 0) {
        System.out.println(in.readInt());
    }
}
```

## 🧾 Uso de `Formatter` e `Scanner`

### 🖊️ Escrita com `Formatter`

#### 1. O que é?

Classe que escreve dados formatados (como `printf`), útil para saída padronizada.

#### 2. Exemplo: gravar dados tabulados

```java
import java.util.Formatter;
import java.io.FileNotFoundException;

public class GravarTabulado {
    public static void main(String[] args) throws FileNotFoundException {
        Formatter fmt = new Formatter("clientes.txt");
        fmt.format("%s %d %.2f\n", "João", 30, 1500.50);
        fmt.format("%s %d %.2f\n", "Maria", 25, 2300.75);
        fmt.close();
    }
}
```

---

### 🔍 Leitura com `Scanner`

#### 1. Exemplo

```java
import java.util.Scanner;
import java.io.File;

public class LerClientes {
    public static void main(String[] args) throws Exception {
        Scanner sc = new Scanner(new File("clientes.txt"));
        while (sc.hasNext()) {
            String nome = sc.next();
            int idade = sc.nextInt();
            double saldo = sc.nextDouble();
            System.out.printf("Nome: %s, Idade: %d, Saldo: %.2f\n", nome, idade, saldo);
        }
        sc.close();
    }
}
```

#### 2. Tratamento de erros

Use `try-catch` com `InputMismatchException` para evitar erros de formatação.

## 🧮 Processamento Sequencial de Arquivos

### 1. Benefício em consulta de crédito

Arquivos sequenciais permitem:

* Consultas rápidas por linha.
* Leitura simples com pouco uso de memória.

### 2. Estrutura ideal

* Registro fixo com campos delimitados.
* Exemplo: número, nome, saldo.

### 3. Exemplo de Implementação

#### a) Gravar dados

```java
import java.io.PrintWriter;

public class GravarClientes {
    public static void main(String[] args) throws Exception {
        try (PrintWriter pw = new PrintWriter("contas.txt")) {
            pw.println("1001 João -150.0");
            pw.println("1002 Maria 500.0");
            pw.println("1003 Ana -75.0");
        }
    }
}
```

#### b) Consulta por conta

```java
import java.util.Scanner;
import java.io.File;

public class ConsultarConta {
    public static void main(String[] args) throws Exception {
        Scanner sc = new Scanner(new File("contas.txt"));
        Scanner entrada = new Scanner(System.in);
        System.out.print("Informe número da conta: ");
        String conta = entrada.next();

        while (sc.hasNextLine()) {
            String linha = sc.nextLine();
            if (linha.startsWith(conta)) {
                System.out.println("Registro: " + linha);
                break;
            }
        }

        sc.close();
    }
}
```

#### c) Listar contas com saldo negativo

```java
Scanner sc = new Scanner(new File("contas.txt"));
while (sc.hasNext()) {
    String num = sc.next();
    String nome = sc.next();
    double saldo = sc.nextDouble();
    if (saldo < 0) {
        System.out.printf("Conta: %s - Nome: %s - Saldo: %.2f\n", num, nome, saldo);
    }
}
sc.close();
```

## 🧬 Serialização de Objetos

### 💡 Conceito e XML

#### 1. O que é Serialização?

Transformar um objeto em uma sequência de bytes para armazenamento ou transmissão.

#### 2. Benefícios do XML

* Leitura por humanos e máquinas.
* Interoperabilidade entre sistemas.

#### 3. Exemplo XML

```xml
<Pessoa>
    <nome>João</nome>
    <idade>30</idade>
</Pessoa>
```

---

### 📦 JAXB - Serializar/Deserializar

#### 1. O que é JAXB?

API Java que converte objetos Java em XML e vice-versa.

#### 2. Classe `Pessoa`

```java
import jakarta.xml.bind.annotation.*;

@XmlRootElement
public class Pessoa {
    private String nome;
    private int idade;

    @XmlElement
    public String getNome() { return nome; }
    public void setNome(String nome) { this.nome = nome; }

    @XmlElement
    public int getIdade() { return idade; }
    public void setIdade(int idade) { this.idade = idade; }
}
```

#### 3. Serializar lista

```java
import jakarta.xml.bind.*;
import java.util.*;

@XmlRootElement
class ListaPessoas {
    private List<Pessoa> pessoas = new ArrayList<>();

    @XmlElement(name = "Pessoa")
    public List<Pessoa> getPessoas() { return pessoas; }
    public void setPessoas(List<Pessoa> pessoas) { this.pessoas = pessoas; }
}

// Serializar
JAXBContext ctx = JAXBContext.newInstance(ListaPessoas.class);
Marshaller marshaller = ctx.createMarshaller();
marshaller.setProperty(Marshaller.JAXB_FORMATTED_OUTPUT, true);
marshaller.marshal(lista, new File("pessoas.xml"));
```

#### 4. Desserializar

```java
Unmarshaller unmarshaller = ctx.createUnmarshaller();
ListaPessoas lista = (ListaPessoas) unmarshaller.unmarshal(new File("pessoas.xml"));
for (Pessoa p : lista.getPessoas()) {
    System.out.println(p.getNome() + " - " + p.getIdade());
}
```
