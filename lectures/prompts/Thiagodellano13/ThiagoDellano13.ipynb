# Coleções Genéricas em Java (Java Generic Collections)

## Introdução às Coleções em Java

### 1. O que são coleções em Java? Por que usamos coleções ao invés de arrays tradicionais?

Coleções são estruturas de dados fornecidas pelo Java Collections Framework que armazenam grupos de objetos.
Diferente dos arrays, que possuem tamanho fixo e funcionalidades limitadas, as coleções oferecem operações como inserção, remoção, busca e ordenação de forma mais eficiente e flexível.

### 2. Quais são as principais interfaces da hierarquia de coleções em Java?

* **List**: mantém a ordem de inserção e permite elementos duplicados (ex: `ArrayList`, `LinkedList`).
* **Set**: não permite elementos duplicados (ex: `HashSet`, `TreeSet`).
* **Map**: associa chaves a valores (ex: `HashMap`, `TreeMap`).

### 3. Exemplo com `ArrayList`:

```java
import java.util.ArrayList;

public class Alunos {
    public static void main(String[] args) {
        ArrayList<String> alunos = new ArrayList<>();
        alunos.add("Ana");
        alunos.add("Bruno");
        alunos.add("Carlos");
        System.out.println(alunos);
    }
}
```

### 4. Vantagens e limitações:

* **Vantagens**: flexibilidade, eficiência, suporte a ordenação, busca, etc.
* **Limitações**: não suportam tipos primitivos diretamente (exigem wrappers), podem consumir mais memória.

## Manipulação de Arrays com a classe `Arrays`

### 1. O que é a classe `Arrays`?

É uma classe utilitária do pacote `java.util` que fornece métodos estáticos para manipulação de arrays (ordenar, preencher, copiar, comparar, etc).

### 2. Exemplos:

```java
import java.util.Arrays;

public class ExemploArrays {
    public static void main(String[] args) {
        int[] numeros = {4, 2, 7, 1};
        Arrays.sort(numeros); // ordena
        Arrays.fill(numeros, 2); // preenche com 2
        int[] copia = Arrays.copyOf(numeros, 6); // copia com tamanho novo
        int indice = Arrays.binarySearch(copia, 2); // busca binária
    }
}
```

### 3. Diferença entre `Arrays.equals()` e `==`

* `==` compara referências de objetos.
* `Arrays.equals()` compara os conteúdos dos arrays.

## Wrapper Classes

### 1. O que são Wrapper Classes?

São classes que encapsulam tipos primitivos para que possam ser usados como objetos (ex: `int` → `Integer`, `double` → `Double`).
Necessárias para usar primitivos em coleções genéricas.

### 2. Exemplo de conversão:

```java
int num = 10;
Integer obj = Integer.valueOf(num); // autoboxing
int prim = obj.intValue(); // unboxing
```

### 3. Programa com `ArrayList<Integer>`:

```java
import java.util.ArrayList;

public class ListaNumeros {
    public static void main(String[] args) {
        ArrayList<Integer> numeros = new ArrayList<>();
        numeros.add(10);
        numeros.add(20);
        numeros.add(30);
        System.out.println(numeros);
    }
}
```

## Autoboxing e Unboxing

### 1. O que são?

* **Autoboxing**: conversão automática de primitivo para wrapper.
* **Unboxing**: conversão de wrapper para primitivo.

### 2. Exemplo com `List<Double>`:

```java
import java.util.List;
import java.util.ArrayList;

public class AutoboxingDemo {
    public static void main(String[] args) {
        List<Double> notas = new ArrayList<>();
        notas.add(7.5); // autoboxing
    }
}
```

### 3. Problemas com unboxing:

```java
Integer nulo = null;
int erro = nulo; // NullPointerException
```

## Uso de Estruturas de Dados Genéricas

### 1. O que são generics?

São recursos que permitem definir tipos parametrizados, aumentando a segurança e reutilização do código.

### 2. Exemplo com `HashSet<String>`:

```java
import java.util.HashSet;

public class ConjuntoNomes {
    public static void main(String[] args) {
        HashSet<String> nomes = new HashSet<>();
        nomes.add("Ana");
        nomes.add("Bruno");
        nomes.add("Ana"); // ignorado
        System.out.println(nomes);
    }
}
```

### 3. Comparações:

```java
ArrayList<String> lista = new ArrayList<>();
LinkedList<String> listaEnc = new LinkedList<>();
HashSet<String> conjunto = new HashSet<>();
TreeSet<String> conjuntoOrdenado = new TreeSet<>();
HashMap<Integer, String> mapa = new HashMap<>();
```

## Algoritmos da classe `Collections`

### 1. O que é a classe `Collections`?

Classe utilitária com métodos estáticos para operações com coleções.

### 2. Exemplos:

```java
import java.util.*;

public class TesteCollections {
    public static void main(String[] args) {
        List<String> palavras = new ArrayList<>(Arrays.asList("sol", "lua", "estrela"));
        Collections.shuffle(palavras);
        Collections.sort(palavras);
        Collections.reverse(palavras);
        System.out.println(Collections.max(palavras));
        System.out.println(Collections.min(palavras));
    }
}
```

### 3. Programa de embaralhar e ordenar:

```java
List<String> palavras = new ArrayList<>(Arrays.asList("azul", "verde", "amarelo"));
Collections.shuffle(palavras);
System.out.println("Embaralhado: " + palavras);
Collections.sort(palavras);
System.out.println("Ordenado: " + palavras);
```

## Iteradores e Navegação em Coleções

### 1. O que é um `Iterator`?

Interface que permite percorrer coleções de forma segura.

### 2. Exemplo com `Iterator`:

```java
List<Integer> numeros = Arrays.asList(1, 2, 3, 4, 5);
Iterator<Integer> it = numeros.iterator();
while (it.hasNext()) {
    int n = it.next();
    if (n % 2 == 0) System.out.println(n);
}
```

### 3. Diferença com `ListIterator`:

```java
List<String> nomes = Arrays.asList("Ana", "Bruno", "Carlos");
ListIterator<String> it = nomes.listIterator(nomes.size());
while (it.hasPrevious()) {
    System.out.println(it.previous());
}
```

## Wrappers de Sincronização e Imutabilidade

### 1. O que é uma coleção sincronizada?

Coleção segura para ambientes com múltiplas threads. Evita condições de corrida.

### 2. Exemplo com `Collections.synchronizedList(...)`:

```java
List<String> lista = Collections.synchronizedList(new ArrayList<>());
```

### 3. Coleções imutáveis:

```java
List<String> imutavel = Collections.unmodifiableList(Arrays.asList("A", "B", "C"));
```

## Fábricas de Coleções Imutáveis no Java 9+

### 1. Métodos `List.of()`, `Set.of()` e `Map.of()`

Criam coleções imutáveis de forma direta e concisa.

### 2. Exemplos:

```java
List<String> lista = List.of("A", "B");
Set<String> conjunto = Set.of("X", "Y");
Map<Integer, String> mapa = Map.of(1, "um", 2, "dois");
```

### 3. Comparações:

* `List.of()` é mais direto e eficiente que `Arrays.asList()`.
* `Collections.unmodifiableList()` exige uma lista existente.

## Projeto Prático

```java
import java.util.*;

public class NotasAlunos {
    public static void main(String[] args) {
        Map<String, List<Double>> turmas = new HashMap<>();

        turmas.put("Ana", Arrays.asList(8.0, 7.5, 9.0));
        turmas.put("Bruno", Arrays.asList(6.0, 5.0, 6.5));
        turmas.put("Carlos", Arrays.asList(9.0, 8.5, 9.5));

        // Média por aluno
        Map<String, Double> medias = new HashMap<>();
        for (Map.Entry<String, List<Double>> entrada : turmas.entrySet()) {
            double media = entrada.getValue().stream().mapToDouble(Double::doubleValue).average().orElse(0);
            medias.put(entrada.getKey(), media);
        }

        // Listar com média > 7
        for (String aluno : medias.keySet()) {
            if (medias.get(aluno) > 7) {
                System.out.println(aluno + ": " + medias.get(aluno));
            }
        }

        // Ordenar por média
        List<Map.Entry<String, Double>> ordenados = new ArrayList<>(medias.entrySet());
        ordenados.sort(Map.Entry.comparingByValue(Comparator.reverseOrder()));

        for (Map.Entry<String, Double> entrada : ordenados) {
            System.out.println(entrada.getKey() + " - Média: " + entrada.getValue());
        }
    }
}
```
